```python
path = r"..\data\laplase_deformation\ConstEqXiDeNoise.csv"
df = pd.read_csv(path, header=0)
xi1 = df["xi1"]
xi2 = df["xi2"]
xi3 = df["xi3"]

r1 = df["r1"]
r2 = df["r2"]
r3 = df["r3"]
```

```python
import pandas as pd
import numpy as np

# Загрузка данных из CSV
data = pd.read_csv(path)

# Функция для проверки условий
def check_conditions(row):
    # Проверка условия по норме
    norm = np.sqrt(row['xi1']**2 + row['xi2']**2 + row['xi3']**2)
    if norm >= 0.01:
        return True  # Строка остается, если норма не выполняется
    
    # Проверка условий по знакам
    sign_xi1_r1 = np.sign(row['xi1']) == np.sign(row['r1'])
    sign_xi2_r2 = np.sign(row['xi2']) == np.sign(row['r2'])
    
    # Если оба условия по знакам выполняются, строка остается
    if sign_xi1_r1 and sign_xi2_r2:
        return True
    
    # Если норма выполняется, но условия по знакам не выполняются, строка удаляется
    return False

# Фильтрация строк
filtered_data = data[data.apply(check_conditions, axis=1)]

# Сохранение отфильтрованных данных в новый CSV-файл
filtered_data.to_csv('Filtered_ConstEqXiDeNoise.csv', index=False)
```
